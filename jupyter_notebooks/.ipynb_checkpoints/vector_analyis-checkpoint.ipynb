# BRB-seq LFC vector analysis

Load the saved freshwater-versus-marine DESeq2 contrasts for gill and liver. Inspect missing and zero LFCs, then calculate vector lengths and angles for every contrast pair within each tissue. No significance filtering is applied.

```r
library(dplyr)
library(tidyr)

repo_root <- "/scratch/ztobias/Gacu_BC_txmics"
notebook_dir <- file.path(repo_root, "jupyter_notebooks")
gill_results <- readRDS(file.path(notebook_dir, "gill_pop_ecotype_contrasts.rds"))
liver_results <- readRDS(file.path(notebook_dir, "liver_pop_ecotype_contrasts.rds"))
```

```r
read_lfc_list <- function(results) {
  if (!is.list(results) || is.null(names(results))) stop("Expected a named list of contrast results")

  out <- lapply(names(results), function(contrast_name) {
    result <- as.data.frame(results[[contrast_name]])
    if (!"log2FoldChange" %in% names(result)) stop("Missing log2FoldChange in contrast: ", contrast_name)
    genes <- rownames(result)
    if (is.null(genes) || anyDuplicated(genes)) stop("Missing or duplicated gene IDs in contrast: ", contrast_name)
    data.frame(gene = genes, lfc = as.numeric(result$log2FoldChange))
  })
  names(out) <- names(results)
  out
}

lfc_results <- list(
  gill = read_lfc_list(gill_results),
  liver = read_lfc_list(liver_results)
)
lapply(lfc_results, names)
```

## Missing and zero LFCs

Exact zero LFCs are retained and counted. Missing or non-finite LFCs are excluded from each pairwise comparison.

```r
lfc_diagnostics <- bind_rows(lapply(names(lfc_results), function(tissue) {
  bind_rows(lapply(names(lfc_results[[tissue]]), function(contrast_name) {
    x <- lfc_results[[tissue]][[contrast_name]]$lfc
    n_finite <- sum(is.finite(x))
    n_zero <- sum(x == 0, na.rm = TRUE)
    data.frame(
      tissue = tissue, contrast = contrast_name, n_genes = length(x),
      n_missing = sum(is.na(x)),
      n_infinite = sum(!is.na(x) & !is.finite(x)),
      n_finite = n_finite, n_zero = n_zero, n_nonzero = n_finite - n_zero,
      pct_missing = 100 * mean(is.na(x)),
      pct_zero_of_finite = if (n_finite > 0) 100 * n_zero / n_finite else NA_real_
    )
  }))
})) %>% arrange(tissue, contrast)

lfc_diagnostics
```

Okay so there are a lot of 0 LFC genes. Let's identity the genes that are 0 across every contrast and drop those.

```r
gill_zero_genes <- lfc_results$gill %>%
  bind_rows(.id = "contrast") %>%
  group_by(gene) %>%
  summarise(all_zero = all(is.finite(lfc) & lfc == 0), .groups = "drop") %>%
  filter(all_zero) %>%
  pull(gene)

liver_zero_genes <- lfc_results$liver %>%
  bind_rows(.id = "contrast") %>%
  group_by(gene) %>%
  summarise(all_zero = all(is.finite(lfc) & lfc == 0), .groups = "drop") %>%
  filter(all_zero) %>%
  pull(gene)
```

```r
length(gill_zero_genes)
length(liver_zero_genes)
```

Now drop them:

```r
lfc_results$gill <- lapply(lfc_results$gill, \(x) filter(x, !gene %in% gill_zero_genes))
lfc_results$liver <- lapply(lfc_results$liver, \(x) filter(x, !gene %in% liver_zero_genes))
```

## Pairwise vector comparisons

Each pair uses genes with finite LFCs in both contrasts; `n_shared_genes` records the overlap. `pearson_angle_deg` matches the legacy notebook's correlation-based angle. `cosine_angle_deg` is the geometric angle from the uncentered LFC vectors.

Remember, this computes across all pairs of pop-eco. Because we saw that extreme difference in overall transcriptional variaitno by year, we will ultimately want to only compare within year. For the marine-benthic-limnetic comparison, we can just include the 2024 solitary lakes as the "general freshwater/lake" adaptation vector.

```r
pairwise_vector_stats <- function(contrast_results, tissue) {
  contrast_names <- names(contrast_results)
  if (length(contrast_names) < 2) return(data.frame())

  pair_idx <- combn(seq_along(contrast_names), 2, simplify = FALSE)
  bind_rows(lapply(pair_idx, function(idx) {
    name_1 <- contrast_names[idx[1]]
    name_2 <- contrast_names[idx[2]]
    shared <- inner_join(contrast_results[[name_1]], contrast_results[[name_2]],
                        by = "gene", suffix = c("_1", "_2")) %>% 
                      filter(is.finite(lfc_1), is.finite(lfc_2))
    x <- shared$lfc_1
    y <- shared$lfc_2
    norm_x <- sqrt(sum(x^2))
    norm_y <- sqrt(sum(y^2))
    cosine <- if (norm_x > 0 && norm_y > 0) sum(x * y) / (norm_x * norm_y) else NA_real_
    correlation <- if (length(x) >= 2 && sd(x) > 0 && sd(y) > 0) cor(x, y) else NA_real_

    data.frame(
      tissue = tissue, contrast_1 = name_1, contrast_2 = name_2,
      n_shared_genes = length(x),
      n_zero_1 = sum(x == 0), n_zero_2 = sum(y == 0),
      n_both_zero = sum(x == 0 & y == 0),
      l2_length_1 = norm_x, l2_length_2 = norm_y,
      pearson_angle_deg = if (!is.na(correlation)) acos(pmax(-1, pmin(1, correlation))) * 180 / pi else NA_real_,
      cosine_angle_deg = if (!is.na(cosine)) acos(pmax(-1, pmin(1, cosine))) * 180 / pi else NA_real_,
      mean_lfc_1 = mean(x),
      mean_lfc_2 = mean(y),
      cosine_similarity = cosine,
      pearson_r = correlation
    )
  }))
}

vector_pairs <- bind_rows(lapply(names(lfc_results), function(tissue) {
  pairwise_vector_stats(lfc_results[[tissue]], tissue)
}))
vector_pairs %>% arrange(tissue, contrast_1, contrast_2)
```

```r
vector_lengths <- bind_rows(
  vector_pairs %>%
    select(tissue, contrast = contrast_1, l2_length = l2_length_1),
  vector_pairs %>%
    select(tissue, contrast = contrast_2, l2_length = l2_length_2)
) %>%
  distinct(tissue, contrast, .keep_all = TRUE) %>%
  arrange(tissue, contrast)
```

```r
make_fw_consensus <- function(lfc_list) {

  # Keep only lake-marine contrasts
  keep <- grepl("_lake_vs_.*_marine$|_marine_vs_.*_lake$", names(lfc_list))
  lfc_list <- lfc_list[keep]

  # Extract population names
  contrast_pops <- strsplit(names(lfc_list), "_vs_", fixed = TRUE)

  freshwater_pop <- vapply(contrast_pops, function(x) {
    sub("_lake$", "", x[grepl("_lake$", x)])
  }, character(1))

  marine_pop <- vapply(contrast_pops, function(x) {
    sub("_marine$", "", x[grepl("_marine$", x)])
  }, character(1))

  # Orient all contrasts as lake - marine
  lfc_oriented <- Map(function(df, name) {
    if (grepl("_marine_vs_", name)) df$lfc <- -df$lfc
    df
  }, lfc_list, names(lfc_list))

  # Unit-normalize vectors
  lfc_unit <- lapply(lfc_oriented, function(df) {
    df$lfc <- df$lfc / sqrt(sum(df$lfc^2))
    df
  })

  # Calculate full and leave-one-out consensuses
  average_vectors <- function(x) {

    groups <- split(seq_along(x), marine_pop)

    lapply(groups, function(idx) {

      dfs <- x[idx]
      pops <- freshwater_pop[idx]
      genes <- dfs[[1]]$gene

      mat <- sapply(dfs, function(df) {
        df$lfc[match(genes, df$gene)]
      })

      make_mean <- function(cols) {
        data.frame(
          gene = genes,
          lfc = rowMeans(mat[, cols, drop = FALSE])
        )
      }

      list(
        all = make_mean(seq_along(dfs)),
        loo = setNames(
          lapply(seq_along(dfs), function(i) {
            make_mean(setdiff(seq_along(dfs), i))
          }),
          pops
        )
      )
    })
  }

  list(
    raw = average_vectors(lfc_oriented),
    unit = average_vectors(lfc_unit)
  )
}

# Gill
consensus_gill <- make_fw_consensus(lfc_results_gill_fm)

# Liver
consensus_liver <- make_fw_consensus(lfc_results_liver_fm)
```

Okay now we'll essentially just paste these onto the existing lfc_results objects and then rerun vector analysis. then we can have this aggregate/summary lake_v_marine vector (done both ways, with unit-scaled and raw). 

```r
flatten_consensus <- function(consensus, method = "raw") {

  x <- consensus[[method]]

  full <- lapply(x, function(marine) marine$all)

  names(full) <- paste0(
    "lake_", method, "_vs_",
    names(x), "_marine"
  )

  loo <- unlist(
    lapply(names(x), function(marine) {

      out <- x[[marine]]$loo

      names(out) <- paste0(
        "lake_", method, "_loo_", names(out),
        "_vs_", marine, "_marine"
      )

      out
    }),
    recursive = FALSE
  )

  c(full, loo)
}
```

Okay so now we have to reorient the vectors again:

```r
consensus_comparisons <- vector_pairs_raw %>%

  # Keep comparisons between one consensus and one population vector
  filter(
    grepl("^lake_raw_", contrast_1) !=
      grepl("^lake_raw_", contrast_2)
  ) %>%

  mutate(
    # Identify consensus and population contrast
    consensus = if_else(
      grepl("^lake_raw_", contrast_1),
      contrast_1, contrast_2
    ),
    contrast = if_else(
      grepl("^lake_raw_", contrast_1),
      contrast_2, contrast_1
    ),

    # Identify reference type
    reference_type = if_else(
      grepl("^lake_raw_loo_", consensus),
      "loo", "full"
    ),

    reference_vector = consensus,

    # Extract marine reference
    marine_pop = sub(".*_vs_", "", consensus),

    marine_actual = case_when(
      grepl("Black Creek_marine", contrast) ~ "Black Creek_marine",
      grepl("Campbell River_marine", contrast) ~ "Campbell River_marine",
      grepl("Sayward_marine", contrast) ~ "Sayward_marine"
    ),

    # Identify freshwater ecotype
    type = case_when(
      grepl("_benthic", contrast) ~ "benthic",
      grepl("_limnetic", contrast) ~ "limnetic",
      grepl("_lake", contrast) ~ "lake",
      TRUE ~ NA_character_
    ),

    # Extract freshwater population regardless of contrast orientation
    freshwater_pop = if_else(
      grepl("_marine_vs_", contrast),
      sub("_(lake|benthic|limnetic)$", "",
          sub(".*_vs_", "", contrast)),
      sub("_(lake|benthic|limnetic)$", "",
          sub("_vs_.*", "", contrast))
    ),

    # Population excluded from LOO consensus
    excluded_pop = if_else(
      reference_type == "loo",
      sub("^lake_raw_loo_(.*?)_vs_.*", "\\1", consensus),
      NA_character_
    ),

    # Correct orientation: freshwater - marine
    angle_corrected = if_else(
      grepl("_marine_vs_", contrast),
      180 - cosine_angle_deg,
      cosine_angle_deg
    ),

    # Length of actual population vector
    l2_length = if_else(
      grepl("^lake_raw_", contrast_1),
      l2_length_2,
      l2_length_1
    )
  ) %>%

  # Retain only correctly matched comparisons
  filter(
    marine_pop == marine_actual,
    !is.na(type),

    # Lakes compared to their own LOO consensus
    (type == "lake" &
       reference_type == "loo" &
       excluded_pop == freshwater_pop) |

    # Benthics and limnetics compared to full consensus
    (type %in% c("benthic", "limnetic") &
       reference_type == "full")
  ) %>%

  # Calculate vector decomposition
  mutate(
    projection = l2_length * cos(angle_corrected * pi / 180),
    perpendicular = l2_length * sin(angle_corrected * pi / 180),

    marine_pop = sub("_marine$", "", marine_pop)
  ) %>%

  dplyr::select(
    tissue,
    marine_pop,
    freshwater_pop,
    type,
    contrast,
    reference_type,
    reference_vector,
    l2_length,
    angle_corrected,
    projection,
    perpendicular
  )
```

Let's get the raw angles between marine-benthic/marine-limnetic and the raw marine-lake.

```r
marine_refs <- c("Black Creek", "Campbell River", "Sayward")

all_vectors <- unique(c(
  vector_pairs_raw$contrast_1,
  vector_pairs_raw$contrast_2
))

ecotype_vectors <- grep(
  "_(benthic|limnetic)_vs_.*_marine$|_marine_vs_.*_(benthic|limnetic)$",
  all_vectors,
  value = TRUE
)

lake_vectors <- grep(
  "_lake_vs_.*_marine$|_marine_vs_.*_lake$",
  all_vectors,
  value = TRUE
)

marine_from_name <- function(x) {
  sub("_marine$", "", stringr::str_extract(
    x, paste0("(", paste(marine_refs, collapse = "|"), ")_marine")
  ))
}

ecotype_lake_angles <- vector_pairs_raw %>%
  filter(
    (contrast_1 %in% ecotype_vectors & contrast_2 %in% lake_vectors) |
    (contrast_2 %in% ecotype_vectors & contrast_1 %in% lake_vectors)
  ) %>%
  mutate(
    ecotype_vector = if_else(
      contrast_1 %in% ecotype_vectors, contrast_1, contrast_2
    ),
    lake_vector = if_else(
      contrast_1 %in% lake_vectors, contrast_1, contrast_2
    ),

    marine_pop = marine_from_name(ecotype_vector),
    lake_marine = marine_from_name(lake_vector),

    # Orient both vectors marine -> freshwater
    sign_ecotype = if_else(grepl("_marine_vs_", ecotype_vector), -1, 1),
    sign_lake = if_else(grepl("_marine_vs_", lake_vector), -1, 1),

    cosine_corrected = cosine_similarity * sign_ecotype * sign_lake,
    angle_corrected = acos(
      pmax(-1, pmin(1, cosine_corrected))
    ) * 180 / pi
  ) %>%
  filter(marine_pop == lake_marine) %>%
  mutate(
    # Extract species-pair lake regardless of contrast orientation
    species_pair = stringr::str_extract(
      ecotype_vector,
      "Little Quarry|Paxton|Priest"
    ),

    ecotype = stringr::str_extract(
      ecotype_vector,
      "benthic|limnetic"
    ),

    # Extract solitary lake regardless of contrast orientation
    solitary_lake = if_else(
      grepl("_marine_vs_", lake_vector),
      sub("_lake$", "", sub(".*_marine_vs_", "", lake_vector)),
      sub("_lake_vs_.*", "", lake_vector)
    )
  ) %>%
  select(
    tissue, marine_pop, species_pair, ecotype,
    solitary_lake, cosine_corrected, angle_corrected
  )

```

```r
make_bl_consensus <- function(lfc_list) {

  # Keep only benthic-limnetic contrasts
  keep <- grepl(
    "_benthic_vs_.*_limnetic$|_limnetic_vs_.*_benthic$",
    names(lfc_list)
  )

  lfc_list <- lfc_list[keep]

  # Keep only within-lake comparisons
  contrast_pops <- strsplit(names(lfc_list), "_vs_", fixed = TRUE)

  same_lake <- vapply(contrast_pops, function(x) {
    sub("_(benthic|limnetic)$", "", x[1]) ==
      sub("_(benthic|limnetic)$", "", x[2])
  }, logical(1))

  lfc_list <- lfc_list[same_lake]

  # Orient all vectors as limnetic - benthic
  lfc_oriented <- Map(function(df, name) {
    if (grepl("_limnetic_vs_", name)) {
      df$lfc <- -df$lfc
    }
    df
  }, lfc_list, names(lfc_list))

  # Calculate mean LFC across the three lakes
  genes <- lfc_oriented[[1]]$gene

  mat <- sapply(lfc_oriented, function(df) {
    df$lfc[match(genes, df$gene)]
  })

  data.frame(
    gene = genes,
    lfc = rowMeans(mat)
  )
}

# Gill
consensus_bl_gill <- make_bl_consensus(lfc_results$gill)

# Liver
consensus_bl_liver <- make_bl_consensus(lfc_results$liver)
```

```r
lake_marine_vectors <- keep_vectors_bl[
  grepl(
    "_lake_vs_.*_marine$|_marine_vs_.*_lake$",
    keep_vectors_bl
  )
]

bl_refs <- c(
  "bl_consensus",
  "Little Quarry_benthic_vs_Little Quarry_limnetic",
  "Paxton_benthic_vs_Paxton_limnetic",
  "Priest_benthic_vs_Priest_limnetic"
)

bl_comparisons <- vector_pairs_bl %>%
  filter(
    (contrast_1 %in% lake_marine_vectors & contrast_2 %in% bl_refs) |
    (contrast_2 %in% lake_marine_vectors & contrast_1 %in% bl_refs)
  )
```

```r
bl_comparisons_plot <- bl_comparisons %>%
  mutate(
    # Identify lake-marine vector and B-L reference
    contrast = if_else(
      contrast_1 %in% lake_marine_vectors,
      contrast_1, contrast_2
    ),

    reference_vector = if_else(
      contrast_1 %in% bl_refs,
      contrast_1, contrast_2
    ),

    # Reference identity
    reference = case_when(
      reference_vector == "bl_consensus" ~ "Consensus",
      grepl("^Paxton_", reference_vector) ~ "Paxton",
      grepl("^Priest_", reference_vector) ~ "Priest",
      grepl("^Little Quarry_", reference_vector) ~ "Little Quarry"
    ),

    # Extract marine reference
    marine_pop = case_when(
      grepl("Black Creek_marine", contrast) ~ "Black Creek",
      grepl("Campbell River_marine", contrast) ~ "Campbell River",
      grepl("Sayward_marine", contrast) ~ "Sayward"
    ),

    # Extract lake identity regardless of contrast orientation
    freshwater_pop = if_else(
      grepl("_marine_vs_", contrast),
      sub("_lake$", "", sub(".*_vs_", "", contrast)),
      sub("_lake$", "", sub("_vs_.*", "", contrast))
    ),

    # Original vector orientations
    lake_reversed = grepl("_marine_vs_", contrast),

    # All four B-L references are currently benthic minus limnetic
    bl_reversed = TRUE,

    # Correct cosine similarity to marine -> lake
    # and benthic -> limnetic orientation
    cosine_corrected = cosine_similarity *
      if_else(lake_reversed, -1, 1) *
      if_else(bl_reversed, -1, 1),

    angle_corrected = acos(
      pmax(-1, pmin(1, cosine_corrected))
    ) * 180 / pi,

    # Length of lake-marine vector
    l2_length = if_else(
      contrast_1 %in% lake_marine_vectors,
      l2_length_1,
      l2_length_2
    ),

    # Projection onto benthic -> limnetic axis
    projection = l2_length * cosine_corrected,

    # Perpendicular magnitude
    perpendicular = l2_length *
      sqrt(pmax(0, 1 - cosine_corrected^2))
  ) %>%
  dplyr::select(
    tissue,
    marine_pop,
    freshwater_pop,
    contrast,
    reference,
    reference_vector,
    l2_length,
    angle_corrected,
    cosine_corrected,
    projection,
    perpendicular
  )
```

```r
library(dplyr)
library(tidyr)

# Prepare data with all three environmental predictors
lake_metrics <- bl_lake_data %>%
  distinct(tissue, marine_pop, freshwater_pop, .keep_all = TRUE) %>%
  mutate(
    log_area = log10(WSA_SURFACE_AREA_HECTARE)
  ) %>%
  pivot_longer(
    cols = c(log_area, MODELLED_MEAN_DEPTH, residual_complexity),
    names_to = "predictor",
    values_to = "x"
  ) %>%
  pivot_longer(
    cols = c(projection, l2_length),
    names_to = "metric",
    values_to = "y"
  ) %>%
  filter(is.finite(x), is.finite(y))

# Correlations separately for each marine reference
lake_correlations <- lake_metrics %>%
  group_by(tissue, marine_pop, predictor, metric) %>%
  summarise(
    n = n(),
    pearson_r = cor(x, y, method = "pearson"),
    pearson_p = cor.test(x, y, method = "pearson")$p.value,
    spearman_rho = cor(x, y, method = "spearman"),
    spearman_p = cor.test(x, y, method = "spearman", exact = FALSE)$p.value,
    r_squared = pearson_r^2,
    .groups = "drop"
  )

lake_correlations


```

```r
paired_consensus <- vector_pairs_raw %>%
  filter(
    grepl("^lake_raw_vs_", contrast_1) |
    grepl("^lake_raw_vs_", contrast_2)
  ) %>%
  mutate(
    consensus = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      contrast_1, contrast_2
    ),
    contrast = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      contrast_2, contrast_1
    ),
    marine_pop = sub("^lake_raw_vs_", "", consensus),

    # Split original contrast into its two populations
    pop_1 = sub("_vs_.*", "", contrast),
    pop_2 = sub(".*_vs_", "", contrast),

    # Identify paired lake
    lake = case_when(
      grepl("Paxton", contrast) ~ "Paxton",
      grepl("Priest", contrast) ~ "Priest",
      grepl("Little Quarry", contrast) ~ "Little Quarry",
      TRUE ~ NA_character_
    ),

    # Classify exact contrast type
    type = case_when(
      grepl("_benthic$", pop_1) & grepl("_limnetic$", pop_2) |
        grepl("_limnetic$", pop_1) & grepl("_benthic$", pop_2) ~
          "benthic-limnetic",

      grepl("_benthic$", pop_1) & grepl("_marine$", pop_2) |
        grepl("_marine$", pop_1) & grepl("_benthic$", pop_2) ~
          "benthic-marine",

      grepl("_limnetic$", pop_1) & grepl("_marine$", pop_2) |
        grepl("_marine$", pop_1) & grepl("_limnetic$", pop_2) ~
          "limnetic-marine",

      TRUE ~ NA_character_
    ),

    # Marine population in the original contrast
    marine_actual = case_when(
      grepl("Black Creek_marine", contrast) ~ "Black Creek",
      grepl("Campbell River_marine", contrast) ~ "Campbell River",
      grepl("Sayward_marine", contrast) ~ "Sayward",
      TRUE ~ NA_character_
    ),

    # Original vector L2 length
    l2_length = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      l2_length_2, l2_length_1
    ),

    # Orient freshwater-marine as freshwater - marine
    # Orient benthic-limnetic as benthic - limnetic
    reversed = case_when(
      type %in% c("benthic-marine", "limnetic-marine") ~
        grepl("_marine$", pop_1),

      type == "benthic-limnetic" ~
        grepl("_limnetic$", pop_1),

      TRUE ~ FALSE
    ),

    angle_corrected = if_else(
      reversed,
      180 - cosine_angle_deg,
      cosine_angle_deg
    )
  ) %>%
  filter(
    !is.na(lake),
    !is.na(type),

    # Marine contrasts must match consensus marine reference;
    # benthic-limnetic contrasts can use all three references
    type == "benthic-limnetic" |
      marine_pop == marine_actual,

    # Benthic-limnetic must come from the SAME lake
    type != "benthic-limnetic" |
      (
        sub("_(benthic|limnetic)$", "", pop_1) ==
        sub("_(benthic|limnetic)$", "", pop_2)
      )
  ) %>%
  mutate(
    projection = l2_length * cos(angle_corrected * pi / 180),
    perpendicular = l2_length * sin(angle_corrected * pi / 180)
  ) %>%
  dplyr::select(
    tissue, marine_pop, lake, type, contrast,
    l2_length, angle_corrected, projection, perpendicular
  ) %>%
  arrange(tissue, marine_pop, lake, type)
```

```r
black_creek_projection <- vector_pairs_raw %>%
  filter(
    grepl("^lake_raw_vs_", contrast_1) |
    grepl("^lake_raw_vs_", contrast_2)
  ) %>%
  mutate(
    consensus = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      contrast_1, contrast_2
    ),
    contrast = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      contrast_2, contrast_1
    ),
    marine_pop = sub("^lake_raw_vs_", "", consensus),

    # Original marine-marine vector length
    l2_length = if_else(
      grepl("^lake_raw_vs_", contrast_1),
      l2_length_2,
      l2_length_1
    )
  ) %>%
  filter(
    # Only Black Creek vs. other marine populations
    contrast %in% c(
      "Black Creek_marine_vs_Campbell River_marine",
      "Black Creek_marine_vs_Sayward_marine"
    ),

    # Compare against freshwater consensus using the OTHER marine
    (grepl("Campbell River_marine$", contrast) &
       marine_pop == "Campbell River") |
    (grepl("Sayward_marine$", contrast) &
       marine_pop == "Sayward")
  ) %>%
  mutate(
    # Both contrasts are already Black Creek - other marine
    angle = cosine_angle_deg,

    projection = l2_length * cos(angle * pi / 180),

    perpendicular = l2_length * sin(angle * pi / 180)
  ) %>%
  dplyr::select(
    tissue, marine_pop, contrast,
    l2_length, angle, projection, perpendicular
  )
```
