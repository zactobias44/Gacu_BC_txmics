# Compare WGS sex assignments with field metadata

This notebook compares the DJR WGS/Y-missingness assignments with field-recorded sex. Samples without a WGS assignment retain the field sex. It also surfaces gravidity, egg, and sex-note information for mismatches and unresolved samples.

```r
library(data.table)

first_existing <- function(paths) paths[file.exists(paths)][1]
norm_id <- function(x) {
  x <- basename(trimws(as.character(x)))
  x <- sub('^ZT_', '', x, ignore.case = TRUE)
  x <- sub('^.*?((2024|2025)_DJR_[0-9]+).*$', '\\1', x, ignore.case = TRUE)
  x
}
norm_sex <- function(x) {
  x <- tolower(trimws(as.character(x)))
  out <- rep(NA_character_, length(x))
  out[grepl('^f|female', x)] <- 'female'
  out[grepl('^m|male', x)] <- 'male'
  out
}

assignment_file <- first_existing(c(
  '../../BC_stickle_expanded_WGS/metadata/derived/wgs_sex_assignment_reconciliation_DJR.tsv',
  '/scratch/ztobias/BC_stickle_expanded_WGS/metadata/derived/wgs_sex_assignment_reconciliation_DJR.tsv'
))
tracking_file <- first_existing(c(
  '../metadata/derived/sample_tracking.tsv',
  '/scratch/ztobias/Gacu_BC_txmics/metadata/derived/sample_tracking.tsv'
))
notes_file <- first_existing(c(
  '../metadata/derived/sample_tracking_notes.tsv',
  '/scratch/ztobias/Gacu_BC_txmics/metadata/derived/sample_tracking_notes.tsv'
))
if (is.na(assignment_file)) stop('DJR WGS assignment table not found.')
if (is.na(tracking_file)) stop('Canonical sample tracking table not found.')

wgs <- fread(assignment_file, data.table = FALSE)
tracking <- fread(tracking_file, data.table = FALSE)
wgs$sample <- norm_id(wgs$sample)
tracking$sample <- norm_id(tracking$sample_id)
wgs <- wgs[grepl('^(2024|2025)_DJR_', wgs$sample), , drop = FALSE]
tracking <- tracking[grepl('^(2024|2025)_DJR_', tracking$sample), , drop = FALSE]

sex_compare <- merge(
  wgs[, c('sample', 'assigned_sex_y_djr', 'y_missing_proportion',
             'assignment_margin_djr', 'assignment_confident_djr')],
  tracking[, c('sample', 'field_sex', 'true_sex', 'population', 'site_type',
               'site_name', 'ecotype', 'wgs_status', 'source_records')],
  by = 'sample', all = TRUE
)
sex_compare$field_sex_normalized <- norm_sex(sex_compare$field_sex)
sex_compare$wgs_sex <- norm_sex(sex_compare$assigned_sex_y_djr)
sex_compare$sex_mismatch <- !is.na(sex_compare$field_sex_normalized) &
  !is.na(sex_compare$wgs_sex) &
  sex_compare$field_sex_normalized != sex_compare$wgs_sex
sex_compare$final_sex <- ifelse(
  !is.na(sex_compare$wgs_sex), sex_compare$wgs_sex,
  sex_compare$field_sex_normalized
)
sex_compare$final_sex_source <- ifelse(
  !is.na(sex_compare$wgs_sex), 'WGS_Y_missingness',
  ifelse(!is.na(sex_compare$field_sex_normalized), 'field', NA_character_)
)

table(sex_compare$sex_mismatch, useNA = 'ifany')
table(sex_compare$final_sex_source, useNA = 'ifany')
```

```r
# Mismatches where both field and WGS assignments exist.
sex_mismatches <- sex_compare[sex_compare$sex_mismatch %in% TRUE, , drop = FALSE]
sex_mismatches[, c('sample', 'field_sex', 'field_sex_normalized',
  'assigned_sex_y_djr', 'y_missing_proportion', 'assignment_confident_djr',
  'population', 'site_name', 'ecotype', 'wgs_status')]

# Samples with no WGS assignment: field sex is carried forward.
field_only <- sex_compare[is.na(sex_compare$wgs_sex) &
  !is.na(sex_compare$field_sex_normalized), , drop = FALSE]
field_only[, c('sample', 'field_sex', 'final_sex', 'final_sex_source',
  'population', 'site_name', 'ecotype', 'wgs_status')]
```

```r
# Bring in original collection notes, including gravidity/egg observations.
collection_files <- c(
  '../../2024_Gacu_BRBseq/metadata/sample_list_2024.csv',
  '/scratch/ztobias/2024_BC_Gaculeatus/sample_list_2024.csv',
  '../metadata/BC Spring 2025 Collections and Molecular Metadata Sheet 1.csv',
  '/scratch/ztobias/Gacu_BC_txmics/metadata/BC Spring 2025 Collections and Molecular Metadata Sheet 1.csv'
)
collection_files <- collection_files[file.exists(collection_files)]
collection_notes <- list()
for (path in collection_files) {
  z <- fread(path, fill = TRUE, data.table = FALSE, encoding = 'UTF-8')
  id_col <- names(z)[tolower(names(z)) %in% c('id', 'sample', 'sample_id')][1]
  if (is.na(id_col)) id_col <- names(z)[1]
  sex_col <- names(z)[tolower(names(z)) %in% c('sex', 'field sex')][1]
  note_candidates <- names(z)[grepl('note|egg|gravid|stain.*sex|sex.*check', names(z), ignore.case = TRUE)]
  if (is.na(sex_col)) sex_col <- NA_character_
  if (!length(note_candidates)) note_candidates <- character(0)
  q <- data.frame(sample = norm_id(z[[id_col]]), stringsAsFactors = FALSE)
  q$field_file_sex <- if (!is.na(sex_col)) z[[sex_col]] else NA
  q$collection_notes <- apply(z[, note_candidates, drop = FALSE], 1, function(v)
    paste(v[!is.na(v) & trimws(v) != ''], collapse = ' | '))
  q$source_file <- basename(path)
  collection_notes[[path]] <- q
}
collection_notes <- if (length(collection_notes)) do.call(rbind, collection_notes) else data.frame()

if (!is.null(notes_file) && !is.na(notes_file)) {
  issue_notes <- fread(notes_file, data.table = FALSE)
  issue_notes$sample <- norm_id(issue_notes$sample_id)
  issue_notes$tracking_note <- paste(issue_notes$issue_type, issue_notes$note,
    issue_notes$recommended_action, sep = ': ')
  issue_notes <- aggregate(tracking_note ~ sample, issue_notes, paste, collapse = ' | ')
} else issue_notes <- data.frame(sample = character(), tracking_note = character())

sex_with_notes <- merge(sex_compare, collection_notes[, c('sample', 'field_file_sex',
  'collection_notes', 'source_file')], by = 'sample', all.x = TRUE)
sex_with_notes <- merge(sex_with_notes, issue_notes, by = 'sample', all.x = TRUE)
sex_with_notes$has_gravid_or_egg_note <- grepl(
  'gravid|egg|ovary|roe',
  paste(sex_with_notes$collection_notes, sex_with_notes$tracking_note),
  ignore.case = TRUE
)

sex_with_notes[sex_with_notes$sex_mismatch | sex_with_notes$has_gravid_or_egg_note,
  c('sample', 'field_sex', 'assigned_sex_y_djr', 'final_sex', 'sex_mismatch',
    'has_gravid_or_egg_note', 'collection_notes', 'tracking_note',
    'population', 'site_name', 'ecotype')]
```

```r
dir.create('../metadata/derived', showWarnings = FALSE, recursive = TRUE)
write.table(sex_with_notes, '../metadata/derived/DJR_sex_field_comparison.tsv',
  sep = '\t', quote = FALSE, row.names = FALSE)
write.table(sex_mismatches, '../metadata/derived/DJR_sex_mismatches.tsv',
  sep = '\t', quote = FALSE, row.names = FALSE)
```
